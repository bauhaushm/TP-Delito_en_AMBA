### 2.2 Carga del dataset con fuente de respaldo y recorte al AMBA

En esta celda cargamos el dataset del SNIC con una función que prueba dos fuentes en orden:

1. **URL oficial del SNIC (Ministerio de Seguridad).** Es la fuente primaria, porque garantiza trabajar con la versión original y más actualizada de los datos.
2. **Copia del recorte ya filtrado (AMBA, 2017-2025) del mismo dataset, almacenada en el repositorio de GitHub del grupo.** Si la página oficial no responde (caída del sitio, error de red o tiempo de espera agotado), la función usa esta copia como respaldo.

Elegimos esta estructura por dos motivos. Primero, **reproducibilidad**: el notebook se ejecuta completo sin intervención manual, tanto en nuestras sesiones de Colab como cuando lo revise otra persona, aunque el servidor del SNIC esté temporalmente fuera de servicio. Segundo, **transparencia**: la función informa por pantalla de cuál de las dos fuentes se cargaron los datos, de modo que queda registro de qué versión se utilizó. Usamos la librería `requests` con un límite de tiempo (`timeout`) para evitar que la ejecución quede detenida indefinidamente esperando una respuesta.

Del archivo original pedimos solo seis columnas: "Año", "Provincia", "Departamento", "Tipo de delito", "Cantidad de hechos" y "Tasa de hechos" (cada 100.000 habitantes). Luego aplicamos dos filtros:

- **Período 2017-2025** (los últimos 9 años), para centrar el análisis en el escenario reciente y reducir el tamaño del dataset.
- **Territorio AMBA:** las 15 comunas de la Ciudad Autónoma de Buenos Aires y los 24 partidos del Gran Buenos Aires (conurbano bonaerense). Para evitar confundir partidos con nombres repetidos en otras provincias (como San Martín o Avellaneda), exigimos que coincidan a la vez la provincia y el departamento.

```python
# 1. Fuentes de datos
URL_SNIC = "https://cloud-snic.minseg.gob.ar/Bases/SNIC/snic-departamentos-anual.xlsx"
URL_REPO = "https://raw.githubusercontent.com/bauhaushm/TP-Delito_en_AMBA/main/Datasets/snic_amba_2017_2025.csv"
columnas = [0, 2, 4, 6, 7, 12]  # Año, Provincia, Departamento, Tipo de delito, Cantidad de hechos, Tasa

def cargar_snic():
    # Intento 1: URL oficial
    try:
        r = requests.get(URL_SNIC, timeout=30)
        r.raise_for_status()
        print("Datos cargados desde la URL oficial del SNIC.")
        return pd.read_excel(io.BytesIO(r.content), usecols=columnas)
    except Exception as e:
        print(f"No se pudo usar la URL oficial ({type(e).__name__}). Usando copia del repositorio...")
    # Intento 2: copia reducida (AMBA, 2017-2025) en el repositorio
    r = requests.get(URL_REPO, timeout=60)
    r.raise_for_status()
    print("Datos cargados desde la copia reducida en el repositorio de GitHub.")
    return pd.read_csv(io.BytesIO(r.content))

# 2. Cargar el dataset nacional
df = cargar_snic()

# 3. Definir el territorio AMBA
partidos_gba = ["Almirante Brown", "Avellaneda", "Berazategui", "Esteban Echeverría",
    "Ezeiza", "Florencio Varela", "General San Martín", "Hurlingham", "Ituzaingó",
    "José C. Paz", "La Matanza", "Lanús", "Lomas de Zamora", "Malvinas Argentinas",
    "Merlo", "Moreno", "Morón", "Quilmes", "San Fernando", "San Isidro", "San Miguel",
    "Tigre", "Tres de Febrero", "Vicente López"]

es_periodo = df["anio"].between(2017, 2025)
es_caba = df["provincia_nombre"] == "Ciudad Autónoma de Buenos Aires"
es_conurbano = (df["provincia_nombre"] == "Buenos Aires") & (df["departamento_nombre"].isin(partidos_gba))

# 4. Filtrar período y territorio
df_amba = df[es_periodo & (es_caba | es_conurbano)].copy().reset_index(drop=True)

# 5. Verificar el resultado
print(f"El dataset del AMBA contiene {len(df_amba)} filas y {len(df_amba.columns)} columnas.")
print("Años incluidos:", sorted(df_amba["anio"].unique().tolist()))

# 6. Guardar el dataset resultante
df_amba.to_csv("snic_amba_2017_2025.csv", index=False)
```

### 2.9 Diagnóstico de los tipos de delito

El SNIC pasa de 56 a 66 tipos de delito en 2023. Identificamos cuáles son los tipos nuevos y comprobamos que no existan categorías agregadas ("Total") que puedan inflar las sumas. Además comparamos los hechos de los tipos comunes y de los nuevos por año, para evaluar si el cambio de clasificación afecta la comparabilidad de la serie.

```python
tipos_por_anio = {a: set(g["codigo_delito_snic_nombre"]) for a, g in df_amba.groupby("anio")}
tipos_comunes = set.intersection(*tipos_por_anio.values())
tipos_nuevos = tipos_por_anio[2025] - tipos_comunes

print(f"Tipos presentes en todos los años: {len(tipos_comunes)}")
print(f"Tipos que aparecen desde 2023: {len(tipos_nuevos)}")
for t in sorted(tipos_nuevos):
    print("  -", t)

print("\nTipos que contienen 'total':", [t for t in df_amba['codigo_delito_snic_nombre'].unique() if "total" in t.lower()])

es_comun = df_amba["codigo_delito_snic_nombre"].isin(tipos_comunes)
resumen = pd.DataFrame({
    "hechos_tipos_comunes": df_amba[es_comun].groupby("anio")["cantidad_hechos"].sum(),
    "hechos_tipos_nuevos": df_amba[~es_comun].groupby("anio")["cantidad_hechos"].sum()
}).fillna(0)
resumen
```

### 3.3 Tasa total de delitos por comuna y partido

Calculamos la tasa total de cada territorio sumando las tasas de todos los tipos de delito (son categorías excluyentes con la misma población de referencia) y promediándola entre 2017 y 2025. Esto permite ordenar a las 39 unidades por nivel de delito relativo a su población.

```python
tasa_total = (df_amba.groupby(["zona", "departamento_nombre", "anio"])["tasa_hechos"]
              .sum().reset_index(name="tasa_total"))
ranking = (tasa_total.groupby(["zona", "departamento_nombre"])["tasa_total"]
           .mean().sort_values(ascending=False).round(1))
ranking
```

Mostramos las cinco comunas o partidos con mayor y menor tasa, y el promedio simple de cada zona:

```python
print(ranking.head(5))
print(ranking.tail(5))
print("\nPromedio simple por zona:")
print(ranking.groupby("zona").mean().round(1))
```

**Interpretación:** al ajustar por población, el panorama se invierte respecto del volumen: aunque el conurbano registra más hechos en total, las comunas de CABA tienen tasas sensiblemente más altas. El promedio simple de la tasa total anual es de unos 5.620 hechos cada 100.000 habitantes en CABA y de unos 2.720 en el conurbano, es decir, algo más del doble. La diferencia es sistemática: la comuna con menor tasa (Comuna 6, cerca de 4.010) supera al partido con mayor tasa (Morón, cerca de 3.730).

Dentro de cada zona también hay dispersión. En CABA encabezan el ranking las comunas 1, 3, 4, 14 y 9, con la Comuna 1 muy por encima del resto (más de 10.000, casi el doble del promedio de CABA). En el conurbano, los partidos con menores tasas son Florencio Varela, La Matanza, Lomas de Zamora, Merlo y San Fernando (entre 2.000 y 2.400), mientras que Morón, General San Martín y Avellaneda se ubican en el extremo alto de la zona. La brecha entre el territorio de mayor y menor tasa es de unas 2,6 veces en CABA y de unas 1,9 en el conurbano.

Estos resultados deben leerse con cautela. La tasa usa como denominador la población residente, por lo que tiende a sobreestimarse en comunas céntricas con mucha población flotante, como la Comuna 1. Además, se trata de delitos registrados, que dependen de la propensión a denunciar y esta puede diferir entre territorios. Por último, los promedios son simples (no ponderados por población). La fuerte heterogeneidad entre comunas justifica profundizar, en la segunda parte del trabajo, en la relación entre el delito y las condiciones socioeconómicas dentro de CABA.

#### 4.1.3 Distribución de la tasa por comuna y partido (boxplot)

```python
# Tasa total promedio 2017-2025 de cada comuna y de cada partido (una observación por unidad)
from matplotlib.ticker import MultipleLocator   # para marcar el eje cada 1.000


# 1) Tasa total por unidad y año: se suman las tasas de todos los tipos de delito de esa comuna o partido
tasa_unidad_anio = (
    df_amba
    .groupby(["zona", "departamento_nombre", "anio"], as_index=False)["tasa_hechos"]
    .sum()
)

# 2) Promedio de los nueve años para cada comuna o partido
tasa_unidad = tasa_unidad_anio.groupby(["zona", "departamento_nombre"])["tasa_hechos"].mean().reset_index()

# 3) Resumen estadístico por zona (para verificar lo que muestra el gráfico)
resumen = tasa_unidad.groupby("zona")["tasa_hechos"].describe()[["count", "min", "25%", "50%", "75%", "max"]].round(0)
print(resumen)

# 4) Boxplot horizontal: una caja por zona, con un punto por cada comuna o partido
orden = ["CABA", "Conurbano"]                     # CABA queda abajo y Conurbano arriba
colores = {"Conurbano": "steelblue", "CABA": "darkorange"}
alto_caja = {"CABA": 0.7, "Conurbano": 0.7}     # alto de cada caja (solo cambia el aspecto, no los datos)
etiquetas = {"CABA": "CABA\n(15 comunas)", "Conurbano": "Conurbano\n(24 partidos)"}
fig, ax = plt.subplots(figsize=(12, 6.5))
azar = np.random.default_rng(0)       # ruido vertical aleatorio, solo para que los puntos no se tapen entre sí
for pos, z in enumerate(orden):
    d = tasa_unidad[tasa_unidad["zona"] == z].sort_values("tasa_hechos")
    caja = ax.boxplot(d["tasa_hechos"], positions=[pos], vert=False, widths=alto_caja[z], patch_artist=True, showfliers=False,
                      medianprops=dict(color="black", linewidth=2), whiskerprops=dict(color="gray"), capprops=dict(color="gray"))
    caja["boxes"][0].set_facecolor(colores[z]); caja["boxes"][0].set_alpha(0.3)
    ax.scatter(d["tasa_hechos"], azar.normal(pos, 0.07 if z == "Conurbano" else 0.05, len(d)), color=colores[z], edgecolor="white", s=55, zorder=3)
    # Nombre del más bajo (debajo) y del más alto (arriba) de cada zona
    ax.annotate(d.iloc[0]["departamento_nombre"], (d.iloc[0]["tasa_hechos"], pos), xytext=(0, -30 if z == "CABA" else -38), textcoords="offset points",
                ha="center", fontsize=9, arrowprops=dict(arrowstyle="-", color="gray", lw=0.7))
    ax.annotate(d.iloc[-1]["departamento_nombre"], (d.iloc[-1]["tasa_hechos"], pos), xytext=(0, 26 if z == "CABA" else 34), textcoords="offset points",
                ha="center", fontsize=9, arrowprops=dict(arrowstyle="-", color="gray", lw=0.7))
    # Nombre del último valor dentro del bigote (el más alto que no es un valor atípico)
    q1, q3 = d["tasa_hechos"].quantile([0.25, 0.75])
    limite = q3 + 1.5 * (q3 - q1)
    ultimo = d[d["tasa_hechos"] <= limite].iloc[-1]
    ax.annotate(ultimo["departamento_nombre"], (ultimo["tasa_hechos"], pos), xytext=(0, -30 if z == "CABA" else -38), textcoords="offset points",
                ha="center", fontsize=9, arrowprops=dict(arrowstyle="-", color="gray", lw=0.7))
    mediana = d["tasa_hechos"].median()
    ax.text(mediana, pos + alto_caja[z] / 2 + 0.08, "mediana " + f"{mediana:,.0f}".replace(",", "."), ha="center", fontsize=9)

ax.set_yticks(range(len(orden))); ax.set_yticklabels([etiquetas[z] for z in orden], fontsize=11)
ax.set_ylim(-0.75, 1.9)
ax.set_title("Distribución de la tasa de delitos por comuna y partido, promedio 2017-2025", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("Hechos cada 100.000 habitantes (promedio 2017-2025)")
ax.set_ylabel("Zona")
ax.xaxis.set_major_formatter(lambda v, _: f"{v:,.0f}".replace(",", "."))
ax.xaxis.set_major_locator(MultipleLocator(1000))      # una marca cada 1.000
ax.grid(axis="x", alpha=0.3)                           # líneas verticales suaves para leer los valores
ax.grid(axis="y", visible=False)
ax.spines[["top", "right"]].set_visible(False)
fig.text(0.01, 0.005, "Fuente: SNIC, Ministerio de Seguridad.", fontsize=8, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("distribucion_tasa_unidades.png", dpi=110)
plt.show()
```

**Interpretación:** las dos zonas se ubican en rangos distintos y casi no se superponen. La mediana de CABA es 5.266 hechos cada 100.000 habitantes y la del conurbano 2.658, cerca de la mitad. La comuna con menor tasa (Comuna 6, 4.012) tiene más que el partido con mayor tasa (Morón, 3.727).

### 4.2 Comparación entre categorías

#### 4.2.1 Composición de los delitos por tipo

Muestra qué porcentaje de los hechos de cada zona corresponde a cada tipo de delito (los 12 más frecuentes y el resto agrupado), para ver qué delitos pesan más en CABA y cuáles en el conurbano. Cada zona suma 100 %, por lo que el gráfico compara la composición y no la cantidad: no dice en cuál de las dos zonas hay más delito.

```python
# 1) Nombres cortos para los tipos de delito más largos (así entran en el gráfico)
nombres_cortos = {
    "Robos (excluye los agravados por el resultado de lesiones y/o muertes)": "Robos",
    "Lesiones culposas en Accidentes Viales": "Lesiones culposas (accidentes viales)",
    "Tentativas de robo (excluye las agravadas por el res. de lesiones y/o muerte)": "Tentativas de robo",
    "Delitos contra la administración pública": "Contra la administración pública",
    "Tenencia simple atenuada para uso personal de estupefacientes": "Tenencia de estupefacientes (uso personal)",
}
# Si el tipo está en el diccionario se usa el nombre corto; si no, se deja el nombre original
df_amba["tipo"] = df_amba["codigo_delito_snic_nombre"].map(nombres_cortos).fillna(df_amba["codigo_delito_snic_nombre"])

# 2) Hechos por tipo de delito y zona (una fila por tipo, una columna por zona)
hechos = df_amba.groupby(["zona", "tipo"])["cantidad_hechos"].sum().unstack(0).fillna(0)

# 3) Porcentaje de cada tipo sobre el total de hechos de su zona
pct = hechos / hechos.sum() * 100

# 4) Nos quedamos con los 12 tipos con mayor suma de porcentajes (entre las dos zonas) y juntamos el resto en una sola fila
top = pct.sum(axis=1).sort_values(ascending=False).head(12).index
tabla_tipos = pct.loc[top]
nombre_otros = f"Otros ({len(pct) - 12} categorías)"          # cuántas categorías se agrupan
tabla_tipos.loc[nombre_otros] = 100 - tabla_tipos.sum()
tabla_tipos = tabla_tipos.iloc[::-1]          # se invierte el orden para que el más frecuente quede arriba
# Control: los porcentajes de cada zona tienen que sumar 100
print("Suma de porcentajes por zona:")
print(tabla_tipos.sum().round(1))

# 5) Gráfico de barras agrupadas: una barra por zona para cada tipo de delito
y = np.arange(len(tabla_tipos))               # posición de cada tipo en el eje vertical
alto = 0.38                                   # alto de cada barra
fig, ax = plt.subplots(figsize=(12, 9))
b1 = ax.barh(y + alto/2, tabla_tipos["Conurbano"], alto, label="Conurbano", color="steelblue")
b2 = ax.barh(y - alto/2, tabla_tipos["CABA"], alto, label="CABA", color="darkorange")
for barras, zona in [(b1, "Conurbano"), (b2, "CABA")]:
    ax.bar_label(barras, labels=[f"{v:.1f}%".replace(".", ",") for v in tabla_tipos[zona]], padding=3, fontsize=9)
ax.set_yticks(y); ax.set_yticklabels(tabla_tipos.index)
ax.set_title("Composición de los delitos denunciados por tipo, 2017-2025", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("Porcentaje del total de hechos de cada zona (%)")
ax.set_ylabel("Tipo de delito")
ax.set_xlim(0, tabla_tipos.max().max() + 6)  # deja lugar a la derecha para las etiquetas
ax.grid(axis="x", alpha=0.3)                                       # grilla solo vertical
ax.set_axisbelow(True)                                             # la grilla queda detrás de los datos
ax.spines[["top", "right"]].set_visible(False)
ax.legend(title="Zona", loc="lower right")
fig.text(0.01, 0.005, "Fuente: SNIC, Ministerio de Seguridad.", fontsize=8, color="gray")
plt.tight_layout(rect=[0, 0.03, 1, 1])
plt.savefig("tipos_delito.png", dpi=110)
plt.show()
```

**Interpretación:** el gráfico muestra qué porcentaje del total de hechos de cada zona corresponde a cada tipo de delito. En las dos zonas los robos y los hurtos son los más frecuentes, pero pesan más en CABA: suman el 59 % de los hechos (32,3 % y 26,7 %), contra el 43,7 % en el conurbano (27,5 % y 16,2 %). También pesan más en CABA otros delitos contra la propiedad (5,8 % contra 3,7 %) y las tentativas de robo (3,4 % contra 1,9 %).

En el conurbano tienen más peso los delitos contra las personas y la seguridad pública: amenazas (13,5 % contra 5,2 % en CABA), lesiones dolosas (10,9 % contra 6,4 %), lesiones culposas por accidentes viales (7,7 % contra 5,2 %), otros delitos contra la seguridad pública (3,2 % contra 1,3 %) y tenencia de estupefacientes para uso personal (2,9 % contra 2,0 %). Las 44 categorías restantes suman el 7,0 % en las dos zonas.

Son porcentajes del total de cada zona, no tasas: que un tipo pese menos en una zona no significa que haya menos casos de ese delito.

#### 4.2.2 Variación de la tasa 2017-2025 por comuna de CABA

Cambio porcentual de la tasa de cada comuna entre 2017 y 2025, ordenado de menor a mayor. Verde indica una baja y rojo un aumento.

```python
# Variación 2017-2025 de la tasa total de cada comuna: se suman las tasas de todos los tipos de delito
# de cada año (dentro de una misma unidad comparten la población) y se compara 2025 con 2017.

# 1) Tasa total por comuna y año, y tabla con una fila por comuna y una columna por año
tasa_anio = (
    df_amba[df_amba["zona"] == "CABA"]
    .groupby(["anio", "departamento_nombre"], as_index=False)["tasa_hechos"]
    .sum()
)
tabla = tasa_anio.pivot(index="departamento_nombre", columns="anio", values="tasa_hechos")
print("Comunas incluidas:", len(tabla))                      # control: tienen que ser 15

# 2) Variación porcentual entre 2017 y 2025, ordenada de menor a mayor (forma de escalera)
variacion = ((tabla[2025] / tabla[2017]) - 1) * 100
variacion = variacion.sort_values()

# 3) Colores: rojo si la tasa subió, verde si bajó (cuanto mayor la variación, más oscuro el tono)
def color(v, tono, v_min, v_max, l_claro=0.58, l_oscuro=0.33):
    t = 0 if v_max == v_min else (abs(v) - v_min) / (v_max - v_min)
    return colorsys.hls_to_rgb(tono, l_claro - t * (l_claro - l_oscuro), 1.0)

pos = variacion[variacion > 0].abs()    # donde la tasa subió
neg = variacion[variacion < 0].abs()    # donde la tasa bajó

colores = [
    color(v, 0.0, pos.min(), pos.max()) if v > 0
    else color(v, 1/3, neg.min(), neg.max())
    for v in variacion
]

# 4) Gráfico de barras
fig, ax = plt.subplots(figsize=(12, 7))
barras = ax.bar(variacion.index, variacion.values, color=colores, width=0.75)
ax.bar_label(barras, labels=[f"{v:+.0f}%".replace("-", "−") for v in variacion], padding=3, fontsize=10)

ax.axhline(0, color="black", linewidth=0.8)                                   # línea del cero
ax.set_title("Variación de la tasa de delitos por comuna de CABA, 2017-2025", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("Comuna")
ax.set_ylabel("Variación porcentual de la tasa (%)")
ax.set_xticks(range(len(variacion)))
ax.set_xticklabels(variacion.index, rotation=55, ha="center")
ax.grid(axis="y", alpha=0.3)                                       # grilla solo horizontal
ax.set_axisbelow(True)                                             # la grilla queda detrás de los datos
ax.spines[["top", "right"]].set_visible(False)
ax.set_ylim(variacion.min() - 6, variacion.max() + 6)              # margen para las etiquetas
fig.text(0.01, 0.008, "Fuente: SNIC, Ministerio de Seguridad.", fontsize=8, color="gray")
fig.tight_layout(rect=[0, 0.04, 1, 1])
plt.savefig("variacion_caba_escalera.png", dpi=110)
plt.show()
```

**Interpretación:** el gráfico muestra cuánto cambió la tasa total de cada comuna de CABA entre 2017 y 2025. Aunque la tasa de CABA en conjunto casi no varió, las comunas tuvieron recorridos distintos: 10 de las 15 bajaron y 5 subieron.

Las mayores bajas fueron las de la Comuna 11 (−17 %), las comunas 2 y 12 (−14 %), la 5 (−13 %) y la 10 (−11 %). Otras cinco bajaron menos, entre −6 % y −1 %. Entre las que subieron, las comunas 3, 13 y 7 casi no cambiaron (entre +1 % y +2 %), mientras que la 8 (+17 %) y la 1 (+23 %) tuvieron aumentos claros.

## 5. Análisis comparativo de las comunas de CABA según zona

La Ciudad de Buenos Aires registra una tasa promedio anual de 5.618,5 hechos cada 100.000 habitantes (2017-2025), pero esa media oculta una dispersión considerable: la Comuna 1 (10.381,2) casi triplica a la Comuna 6 (4.011,6). Además, el delito no se ordena según el nivel de ingresos: las comunas con mayores tasas (1, 3, 4, 14 y 9) pertenecen a las tres zonas de la Ciudad. Esto sugiere que el delito registrado responde a más de un mecanismo, por lo que conviene analizar cada zona por separado.



```python
# =============================================================================
# INDICADORES SOCIOECONÓMICOS POR COMUNA (EAH - IDECBA): carga, limpieza y base
# =============================================================================
import glob
import logging
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import seaborn as sns

# Silencia solo los avisos internos de openpyxl; los de seaborn/pandas siguen visibles
warnings.filterwarnings("ignore", module="openpyxl")
logging.getLogger("openpyxl").setLevel(logging.ERROR)
sns.set_theme(style="whitegrid", context="notebook")

# ---------------------------------------------------------------------------
# 0. PARÁMETROS (editables)
# ---------------------------------------------------------------------------
ANIOS = range(2017, 2026)    # años a procesar (2020 se omite solo: sin datos por comuna)
ANIO_GRAFICO = 2024          # año que se muestra en los gráficos

# Las mismas 3 zonas y colores que en la parte de condiciones territoriales
PERFILES = {"Norte": [2, 11, 12, 13, 14],
            "Centro": [1, 3, 5, 6, 7, 10, 15],
            "Sur": [4, 8, 9]}
ORDEN_PERFILES = ["Norte", "Centro", "Sur"]
PALETA = {"Norte": "#EC6A9C", "Centro": "#24A57A", "Sur": "#3A8FE0"}


# ---------------------------------------------------------------------------
# 1. DETECCIÓN AUTOMÁTICA DE ARCHIVOS
# ---------------------------------------------------------------------------
# Leemos los Excel de la EAH directamente desde el GitHub del grupo
url_datos = "https://raw.githubusercontent.com/bauhaushm/TP-Delito_en_AMBA/main/Datasets/"
file_ipcf = url_datos + "MT_eah_2417.xlsx"  # Ingreso per cápita familiar
file_emp = url_datos + "MT_eah_0917.xlsx"   # Actividad, empleo, desocupación
file_h04 = url_datos + "H04_co.xlsx"        # Personas por ambiente (hacinamiento)
print(f"✓ Ingresos: {file_ipcf}\n✓ Empleo: {file_emp}\n✓ Vivienda: {file_h04}\n")


# ---------------------------------------------------------------------------
# 2. FUNCIONES DE EXTRACCIÓN
# ---------------------------------------------------------------------------
def filas_comunas(raw, columnas, nombres):
    """Filas de las comunas 1-15 (se identifican por su valor, no por su posición).
    Las hojas de empleo traen bloques Total / Varones / Mujeres: se conserva el primero."""
    num = pd.to_numeric(raw[0], errors="coerce")
    out = raw[num.between(1, 15)].iloc[:, columnas].copy()
    out.columns = nombres
    out = out.apply(pd.to_numeric, errors="coerce")   # '---' (dato suprimido) -> NaN
    return out.drop_duplicates("Comuna", keep="first")


def valor_total(raw, columna):
    """Valor de la fila 'Total' (CABA completa) en la columna indicada."""
    fila = raw[raw[0].astype(str).str.strip() == "Total"].iloc[0]
    return pd.to_numeric(fila[columna], errors="coerce")


def procesar_anio(anio):
    hoja = str(anio)

    # A. Ingreso per cápita familiar (pesos corrientes del año)
    raw = pd.read_excel(file_ipcf, sheet_name=hoja, header=None)
    ipcf = filas_comunas(raw, [0, 1], ["Comuna", "IPCF_Promedio"])
    ipcf_total = valor_total(raw, 1)

    # B. Mercado de trabajo
    raw = pd.read_excel(file_emp, sheet_name=hoja, header=None)
    emp = filas_comunas(raw, [0, 1, 2, 3, 5],
                        ["Comuna", "Tasa_Actividad", "Tasa_Empleo",
                         "Tasa_Desocupacion", "Tasa_Subocupacion"])
    desoc_total = valor_total(raw, 3)

    # C. Hacinamiento: % de hogares con MÁS de 2 personas por ambiente.
    # Se calcula como 100 - (hasta 0,99) - (1 a 1,99). Sumar directamente las categorías
    # "2-3" y "más de 3" subestima el valor, porque el IDECBA suprime ('---') o marca '-'
    # las celdas con pocos casos, y esas quedaban como NaN o como 0.
    raw = pd.read_excel(file_h04, sheet_name=hoja, header=None)
    hac = filas_comunas(raw, [0, 2, 3], ["Comuna", "hasta_0_99", "de_1_a_1_99"])
    hac["Pct_Hacinamiento"] = 100 - hac["hasta_0_99"] - hac["de_1_a_1_99"]

    df = (ipcf.merge(emp, on="Comuna").merge(hac[["Comuna", "Pct_Hacinamiento"]], on="Comuna"))
    if df.empty:
        return df
    df["Comuna"] = df["Comuna"].astype(int)
    df["IPCF_Total_CABA"] = ipcf_total
    # Los pesos de distintos años no son comparables (inflación): índice relativo, CABA = 100
    df["IPCF_Relativo"] = (df["IPCF_Promedio"] / ipcf_total * 100).round(2)   # hasta 2 decimales
    df["Desocupacion_Total_CABA"] = desoc_total
    df["Año"] = anio
    return df


# ---------------------------------------------------------------------------
# 3. EJECUCIÓN Y GUARDADO
# ---------------------------------------------------------------------------
bloques = []
for anio in ANIOS:
    try:
        df_anio = procesar_anio(anio)
    except ValueError:   # la hoja del año no existe en alguno de los archivos
        print(f"⚠️ {anio}: no existe la hoja en alguno de los archivos. Se omite.")
        continue
    if df_anio.empty:
        print(f"⚠️ {anio}: los archivos no traen datos por comuna para ese año. Se omite.")
        continue
    bloques.append(df_anio)
    print(f"✓ Año {anio} procesado correctamente.")

df_socioeconomico_caba = pd.concat(bloques, ignore_index=True)
anio_min, anio_max = df_socioeconomico_caba["Año"].min(), df_socioeconomico_caba["Año"].max()
archivo_salida = f"socioeconomico_caba_{anio_min}_{anio_max}.csv"
df_socioeconomico_caba.to_csv(archivo_salida, index=False)
print(f"\nArchivo '{archivo_salida}' generado con éxito. Filas: {len(df_socioeconomico_caba)}")
nulos = df_socioeconomico_caba.isna().sum()
nulos = nulos[nulos > 0]
print("Valores nulos:", "ninguno" if nulos.empty else nulos.to_dict())

# Vista previa con formato de pesos: se hace sobre una COPIA para no convertir
# la columna numérica en texto (eso es lo que rompía los gráficos).
vista = df_socioeconomico_caba.copy()
vista["IPCF_Promedio"] = vista["IPCF_Promedio"].apply(
    lambda x: f"$ {int(x):,}".replace(",", ".") if pd.notna(x) else "N/D")
print("\nVista previa del dataset consolidado:")
display(vista.head(15))


# ---------------------------------------------------------------------------
# 4. APOYO PARA LOS GRÁFICOS
# ---------------------------------------------------------------------------
def asignar_perfil(c):
    for perfil, comunas in PERFILES.items():
        if c in comunas:
            return perfil
    return "Resto de CABA"


def fmt_pesos(v):
    return "$ " + f"{v:,.0f}".replace(",", ".")


def fmt_pct(v):
    return f"{v:.1f}".replace(".", ",") + " %"


def barras_por_comuna(columna, titulo, xlabel, total_caba, fmt, nota):
    """Barras horizontales ordenadas, coloreadas por perfil, con línea del total CABA."""
    d = df_socioeconomico_caba[df_socioeconomico_caba["Año"] == ANIO_GRAFICO].copy()
    d["Perfil"] = d["Comuna"].apply(asignar_perfil)
    d["Etiqueta"] = "Comuna " + d["Comuna"].astype(str)
    d = d.sort_values(columna, ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(10, 7.5))
    sns.barplot(data=d, y="Etiqueta", x=columna, hue="Perfil", hue_order=ORDEN_PERFILES,
                palette=PALETA, dodge=False, order=d["Etiqueta"], ax=ax)
    tope = d[columna].max()
    for i, v in enumerate(d[columna]):                  # valor al final de cada barra
        ax.text(v + tope * 0.012, i, fmt(v), va="center", fontsize=9, zorder=5,
                bbox=dict(boxstyle="square,pad=0.1", fc="white", ec="none"))
    ax.axvline(total_caba, color="black", linestyle="--", linewidth=1.2)
    ax.set_title(titulo, fontsize=12, fontweight="bold")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Comuna")
    ax.set_xlim(0, tope * 1.15)
    handles = [Patch(facecolor=PALETA[p], label=p) for p in ORDEN_PERFILES]
    handles.append(Line2D([0], [0], color="black", linestyle="--", linewidth=1.2,
                          label=f"Total CABA: {fmt(total_caba)}"))
    ax.legend(handles=handles, title="Zona", loc="lower right", frameon=True)
    fig.text(0.01, -0.01, nota, fontsize=8.5, color="#555555")
    plt.tight_layout()
    return fig, ax


def total_del_anio(columna_total):
    return df_socioeconomico_caba.loc[df_socioeconomico_caba["Año"] == ANIO_GRAFICO,
                                      columna_total].iloc[0]
```

### 5.1 Ingreso per cápita familiar por comuna

- 🟢 **Ingresos altos:** 10 % o más por encima del promedio.
- 🟡 **Ingresos medios:** hasta 10 % por encima o por debajo del promedio.
- 🔴 **Ingresos bajos:** 10 % o más por debajo del promedio.

Usamos solo el nivel de ingreso, sin agrupar las comunas por zona.

```python

# 1. Nos quedamos con el año del gráfico y ordenamos de mayor a menor ingreso
d = df_socioeconomico_caba[df_socioeconomico_caba["Año"] == ANIO_GRAFICO].copy()
d = d.sort_values("IPCF_Promedio")
d["Etiqueta"] = "Comuna " + d["Comuna"].astype(str)

# 2. Clasificamos cada comuna comparando su ingreso con el promedio de CABA (= 100)
def semaforo(valor):
    if valor >= 110:
        return "Ingresos altos"
    elif valor >= 90:
        return "Ingresos medios"
    else:
        return "Ingresos bajos"

d["Nivel"] = d["IPCF_Relativo"].apply(semaforo)

# 3. Un color para cada nivel
colores = {"Ingresos altos": "#2E9E5B", "Ingresos medios": "#F2C14E", "Ingresos bajos": "#D64545"}

# 4. Gráfico de barras horizontales
fig, ax = plt.subplots(figsize=(10, 7.5))
barras = ax.barh(d["Etiqueta"], d["IPCF_Promedio"], color=d["Nivel"].map(colores))
ax.bar_label(barras, labels=d["IPCF_Promedio"].apply(fmt_pesos), padding=4, fontsize=9,
             bbox=dict(facecolor="white", edgecolor="none", pad=1))

# Línea punteada con el promedio de CABA
total = total_del_anio("IPCF_Total_CABA")
ax.axvline(total, color="black", linestyle="--", linewidth=1.2)

ax.set_title(f"Ingreso per cápita familiar (IPCF) promedio por comuna\n(CABA, {ANIO_GRAFICO})",
             fontsize=12, fontweight="bold")
ax.set_xlabel(f"IPCF promedio (pesos corrientes de {ANIO_GRAFICO})")
ax.set_ylabel("Comuna")
ax.set_xlim(0, d["IPCF_Promedio"].max() * 1.18)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f"{v:,.0f}".replace(",", ".")))
ax.grid(axis="y", visible=False)

# Leyenda
leyenda = [Patch(color="#2E9E5B", label="Ingresos altos (10% o más sobre el promedio)"),
           Patch(color="#F2C14E", label="Ingresos medios (cerca del promedio, ±10%)"),
           Patch(color="#D64545", label="Ingresos bajos (10% o más debajo del promedio)"),
           Line2D([0], [0], color="black", linestyle="--", label=f"Promedio CABA: {fmt_pesos(total)}")]
ax.legend(handles=leyenda, loc="upper center", bbox_to_anchor=(0.5, -0.1), ncol=2, frameon=False)

fig.text(0.01, -0.09, f"Fuente: elaboración propia con datos de la EAH {ANIO_GRAFICO} (IDECBA). "
         "Pesos corrientes: solo comparables entre comunas del mismo año.", fontsize=8.5, color="#555555")
plt.tight_layout()
plt.savefig("grafico_ipcf_por_comuna.png", dpi=300, bbox_inches="tight")
plt.show()
```

La brecha entre extremos es amplia: el IPCF de la Comuna 2 ($ 1.046.451) es 2,6 veces el de la Comuna 8 ($ 401.724), que apenas alcanza la mitad del promedio de la Ciudad.

El grupo rojo no es homogéneo. Las comunas 8, 4 y 9 combinan bajos ingresos con las peores condiciones habitacionales, como se verá en la sección de condiciones territoriales, mientras que las comunas 1 y 3 quedan por debajo del promedio pero registran las tasas de delito más altas. Las estimaciones por comuna tienen error muestral, por lo que las diferencias pequeñas deben leerse con cautela.

Calculamos la tasa de delito anual de cada comuna (SNIC) y la unimos con los indicadores de la EAH, promediando los años con datos en ambas fuentes, para usarla en los gráficos siguientes:

```python
# =============================================================================
# ANÁLISIS COMPARATIVO DEL DELITO EN LAS COMUNAS DE CABA (2017-2025)
# Gráficos: (a) Delito vs. Ingreso, (b) Delito vs. Hacinamiento/Desocupación,
#           (c) Ranking de comunas por Zona
# =============================================================================

# ---------------------------------------------------------------------------
# 0. LIBRERÍAS Y CONFIGURACIÓN
# ---------------------------------------------------------------------------
import logging
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.patches import Patch
import seaborn as sns

# Silencia únicamente los avisos internos de openpyxl (no los de seaborn/pandas)
warnings.filterwarnings("ignore", module="openpyxl")
logging.getLogger("openpyxl").setLevel(logging.ERROR)

sns.set_theme(style="whitegrid", context="notebook")


# Las mismas 3 zonas y colores que en la parte de condiciones territoriales
PERFILES = {"Norte": [2, 11, 12, 13, 14],
            "Centro": [1, 3, 5, 6, 7, 10, 15],
            "Sur": [4, 8, 9]}
ORDEN_PERFILES = ["Norte", "Centro", "Sur"]
PALETA = {"Norte": "#EC6A9C", "Centro": "#24A57A", "Sur": "#3A8FE0"}

COMUNAS_ETIQUETADAS = [1, 4, 8, 14]   # comunas rotuladas en los scatter
ANIOS_SOCIO = range(2017, 2026)       # la EAH no se relevó en 2020 (se omite sola)


def formato_miles(x, _pos=None):
    """12345 -> '12.345' (separador de miles con punto, como en Argentina)."""
    return f"{x:,.0f}".replace(",", ".")


# ---------------------------------------------------------------------------
# 1. DELITO (SNIC): tasa total anual por comuna
# ---------------------------------------------------------------------------
snic = pd.read_csv("https://raw.githubusercontent.com/bauhaushm/TP-Delito_en_AMBA/main/Datasets/snic_amba_2017_2025.csv")
snic = snic[
    (snic["provincia_nombre"] == "Ciudad Autónoma de Buenos Aires")
    & (snic["departamento_nombre"].str.startswith("Comuna"))   # excluye "sin determinar"
].copy()

# Serie comparable: solo los tipos de delito presentes en todos los años
# (el SNIC amplía el listado de tipos de delito a partir de 2023)
tipos_por_anio = snic.groupby("anio")["codigo_delito_snic_nombre"].apply(set)
tipos_comunes = set.intersection(*tipos_por_anio.values)
snic = snic[snic["codigo_delito_snic_nombre"].isin(tipos_comunes)]

# Tasa total = suma de las tasas de cada tipo (categorías excluyentes, misma población)
tasa_anual = (snic.groupby(["departamento_nombre", "anio"])["tasa_hechos"]
              .sum().reset_index(name="Tasa_Delito"))
tasa_anual["Comuna"] = tasa_anual["departamento_nombre"].str.extract(r"(\d+)")[0].astype(int)
tasa_anual = tasa_anual.drop(columns="departamento_nombre")

# Promedio 2017-2025 (todos los años): es el que se usa en el ranking del gráfico c
ranking_caba = tasa_anual.groupby("Comuna")["Tasa_Delito"].mean().reset_index()
print(f"Tasa promedio CABA 2017-2025: {ranking_caba['Tasa_Delito'].mean():,.1f} por 100.000 hab.")


# ---------------------------------------------------------------------------
# 2. INDICADORES SOCIOECONÓMICOS (EAH): ingreso, desocupación, hacinamiento
# ---------------------------------------------------------------------------
url_datos = "https://raw.githubusercontent.com/bauhaushm/TP-Delito_en_AMBA/main/Datasets/"
f_ipcf = url_datos + "MT_eah_2417.xlsx"   # Ingreso per cápita familiar
f_emp = url_datos + "MT_eah_0917.xlsx"    # Mercado de trabajo
f_h04 = url_datos + "H04_co.xlsx"         # Personas por ambiente

def solo_comunas(df, cols):
    """Se queda con las filas de comunas 1-15 (descarta títulos, 'Total' y notas)."""
    num = pd.to_numeric(df[0], errors="coerce")
    out = df[num.between(1, 15)].iloc[:, cols].copy()
    out.columns = ["Comuna"] + list(out.columns[1:])
    return out


def leer_anio(anio):
    hoja = str(anio)

    # --- Ingreso per cápita familiar (IPCF), en pesos corrientes del año ---
    raw = pd.read_excel(f_ipcf, sheet_name=hoja, header=None)
    ipcf_total_caba = pd.to_numeric(
        raw.loc[raw[0].astype(str).str.strip() == "Total", 1].iloc[0])
    ipcf = solo_comunas(raw, [0, 1]); ipcf.columns = ["Comuna", "IPCF"]

    # --- Desocupación (la hoja trae Total / Varones / Mujeres: nos quedamos con el Total) ---
    raw = pd.read_excel(f_emp, sheet_name=hoja, header=None)
    emp = solo_comunas(raw, [0, 3]); emp.columns = ["Comuna", "Desocupacion"]
    emp = emp.drop_duplicates("Comuna", keep="first")

    # --- Hacinamiento: % de hogares con MÁS de 2 personas por ambiente ---
    # = 100 - (hasta 0,99) - (1 a 1,99). Evita las celdas suprimidas ('---') de las
    # categorías pequeñas (CV > 30%), que son las que rompen la suma directa.
    raw = pd.read_excel(f_h04, sheet_name=hoja, header=None)
    hac = solo_comunas(raw, [0, 2, 3]); hac.columns = ["Comuna", "p1", "p2"]
    for c in ("p1", "p2"):
        hac[c] = pd.to_numeric(hac[c], errors="coerce")
    hac["Hacinamiento"] = 100 - hac["p1"] - hac["p2"]

    df = (ipcf.merge(emp, on="Comuna").merge(hac[["Comuna", "Hacinamiento"]], on="Comuna"))
    df = df.apply(pd.to_numeric, errors="coerce").astype(float)   # todo numérico (float)
    df["Comuna"] = df["Comuna"].astype(int)
    # IMPORTANTE: el IPCF está en pesos corrientes y hubo mucha inflación entre 2017 y 2025.
    # Promediar pesos nominales de distintos años no tiene sentido. Se usa un ÍNDICE
    # RELATIVO (promedio de CABA = 100 en cada año), que es comparable entre años.
    df["IPCF_Relativo"] = df["IPCF"] / ipcf_total_caba * 100
    df["anio"] = anio
    return df


socio_anual = []
for a in ANIOS_SOCIO:
    try:
        socio_anual.append(leer_anio(a))
    except ValueError:      # hoja inexistente (p. ej. 2020)
        print(f"Aviso: sin datos EAH para {a}; se omite ese año.")
socio_anual = pd.concat(socio_anual, ignore_index=True)


# ---------------------------------------------------------------------------
# 3. BASE FINAL POR COMUNA (promedio de los años con datos EAH)
# ---------------------------------------------------------------------------
# Para los scatter se usan los MISMOS años en delito y en variables socioeconómicas.
panel = socio_anual.merge(tasa_anual, on=["Comuna", "anio"], how="inner")
VARS = ["Tasa_Delito", "IPCF_Relativo", "Desocupacion", "Hacinamiento"]
base = panel.groupby("Comuna")[VARS].mean().astype(float).reset_index()


def asignar_perfil(c):
    for perfil, comunas in PERFILES.items():
        if c in comunas:
            return perfil
    return "Resto de CABA"


base["Perfil"] = base["Comuna"].apply(asignar_perfil)
ranking_caba["Perfil"] = ranking_caba["Comuna"].apply(asignar_perfil)

print("\nAños usados en los scatter:", sorted(int(a) for a in panel["anio"].unique()))
print(base.round(1).to_string(index=False))

# ---------------------------------------------------------------------------
# 4. FUNCIONES AUXILIARES DE GRAFICACIÓN
# ---------------------------------------------------------------------------
def scatter_con_tendencia(ax, x, xlabel, titulo):
    """Dispersión por perfil + recta de tendencia (mínimos cuadrados) + rótulos."""
    sns.regplot(data=base, x=x, y="Tasa_Delito", scatter=False, ci=None,
                color="#444444", line_kws={"linewidth": 1.8, "linestyle": "--"}, ax=ax)
    sns.scatterplot(data=base, x=x, y="Tasa_Delito", hue="Perfil", hue_order=ORDEN_PERFILES,
                    palette=PALETA, s=130, edgecolor="white", linewidth=0.8,
                    legend=False, ax=ax, zorder=3)
    for _, fila in base[base["Comuna"].isin(COMUNAS_ETIQUETADAS)].iterrows():
        ax.annotate(f"Comuna {int(fila['Comuna'])}", (fila[x], fila["Tasa_Delito"]),
                    xytext=(8, 8), textcoords="offset points", fontsize=10, fontweight="bold")
    ax.set_title(titulo, fontsize=12, fontweight="bold")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Tasa de delito (hechos cada 100.000 hab.)")
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(formato_miles))


def leyenda_perfiles(fig_o_ax, **kw):
    handles = [Patch(facecolor=PALETA[p], label=p) for p in ORDEN_PERFILES]
    return fig_o_ax.legend(handles=handles, title="Zona", frameon=True, **kw)
```

### 5.3 Delito e ingreso per cápita familiar

Cada punto es una comuna: su posición muestra su nivel de ingreso (promedio de CABA = 100) y su tasa de delito, ambos promediados entre 2017 y 2025, sin 2020, porque la EAH no relevó ese año por comuna. La línea punteada indica la tendencia general.

```python
# ---------------------------------------------------------------------------
# 5. GRÁFICO (a): DELITO vs. INGRESO PER CÁPITA FAMILIAR
# ---------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6.5))
scatter_con_tendencia(
    ax, "IPCF_Relativo",
    "Ingreso per cápita familiar relativo (promedio CABA = 100)",
    "Tasa de delito e ingreso per cápita familiar por comuna\n(CABA, promedio de años con datos EAH 2017-2025)")
leyenda_perfiles(ax, loc="upper right")
fig.text(0.01, -0.02, "Fuente: elaboración propia con datos del SNIC y de la EAH (IDECBA). "
         "La línea punteada es el ajuste lineal (mínimos cuadrados). Unidad de análisis: comuna (n = 15).",
         fontsize=8.5, color="#555555")
plt.tight_layout()
plt.savefig("grafico_a_delito_vs_ingreso.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Interpretación:** la línea de tendencia baja levemente: las comunas con más ingresos tienden a tener algo menos de delito, pero la relación es débil y los puntos están muy dispersos. Los casos que más se apartan son claros: la Comuna 1, con ingresos cercanos al promedio, tiene la tasa más alta de la Ciudad; la Comuna 8, la de menores ingresos, queda por debajo del promedio de delito; y la Comuna 14, la de mayores ingresos, lo supera. El ingreso, por sí solo, no explica cuánto delito se registra en cada comuna.

### 5.4 Delito, hacinamiento y desocupación

Repetimos la comparación con dos condiciones de vida: el porcentaje de hogares hacinados (más de 2 personas por ambiente) y la tasa de desocupación.

```python
# ---------------------------------------------------------------------------
# 6. GRÁFICO (b): DELITO vs. HACINAMIENTO Y DESOCUPACIÓN
# ---------------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(15, 6), sharey=True)
scatter_con_tendencia(axes[0], "Hacinamiento",
                      "Hogares con más de 2 personas por ambiente (%)",
                      "Delito y hacinamiento")
scatter_con_tendencia(axes[1], "Desocupacion",
                      "Tasa de desocupación (%)", "Delito y desocupación")
axes[1].set_ylabel("")
leyenda_perfiles(fig, loc="lower center", ncol=4, bbox_to_anchor=(0.5, -0.07))
fig.suptitle("Tasa de delito y condiciones de vida por comuna (CABA)",
             fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.savefig("grafico_b_delito_vs_hacinamiento_desocupacion.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Interpretación:** el hacinamiento muestra una relación más clara con el delito que el ingreso: las comunas con más hogares hacinados (1, 8 y 4) están entre las de mayor tasa, aunque buena parte de esa relación se debe a la Comuna 1, y al excluirla se debilita. En el caso de la desocupación, la relación es débil: la Comuna 8 tiene la desocupación más alta pero una tasa de delito por debajo del promedio. En ambos casos, las condiciones de vida acompañan las diferencias entre comunas, pero no alcanzan para explicarlas.

### 5.5 Ranking de comunas por tasa de delito

Ordenamos las 15 comunas según su tasa promedio anual de delito (2017-2025), coloreadas por zona de la Ciudad.

```python
# ---------------------------------------------------------------------------
# 7. GRÁFICO (c): RANKING DE COMUNAS POR TASA DE DELITO (2017-2025)
# ---------------------------------------------------------------------------
rk = ranking_caba.sort_values("Tasa_Delito", ascending=False).reset_index(drop=True)
rk["Etiqueta"] = "Comuna " + rk["Comuna"].astype(str)
promedio_caba = rk["Tasa_Delito"].mean()

fig, ax = plt.subplots(figsize=(10, 7.5))
sns.barplot(data=rk, y="Etiqueta", x="Tasa_Delito", hue="Perfil", hue_order=ORDEN_PERFILES,
            palette=PALETA, dodge=False, order=rk["Etiqueta"], ax=ax)
for i, v in enumerate(rk["Tasa_Delito"]):                 # valor al final de cada barra
    ax.text(v + 120, i, formato_miles(v), va="center", fontsize=9, zorder=5,
            bbox=dict(boxstyle="square,pad=0.1", fc="white", ec="none"))
ax.axvline(promedio_caba, color="black", linestyle="--", linewidth=1.2)
ax.text(promedio_caba + 100, len(rk) - 0.6, f"Promedio CABA: {formato_miles(promedio_caba)}",
        fontsize=9, va="bottom", zorder=5)
ax.set_title("Ranking de comunas por tasa promedio anual de delito\n(CABA, 2017-2025)",
             fontsize=12, fontweight="bold")
ax.set_xlabel("Tasa de delito (hechos cada 100.000 hab.)")
ax.set_ylabel("Comuna")
ax.set_xlim(0, rk["Tasa_Delito"].max() * 1.12)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(formato_miles))
ax.legend(title="Zona", loc="lower right", frameon=True)
fig.text(0.01, -0.01, f"Fuente: elaboración propia con datos del SNIC (56 tipos de delito "
         "comparables en todo el período; excluye hechos sin comuna asignada).", fontsize=8.5, color="#555555")
plt.tight_layout()
plt.savefig("grafico_c_ranking_comunas.png", dpi=300, bbox_inches="tight")
plt.show()
```

### 5.6 Interpretación por zona
   
**Zona Centro (Comunas 1, 3, 5, 6, 7, 10 y 15).** Es la zona más heterogénea. Incluye las dos comunas con las tasas más altas de la Ciudad, la 1 (10.381,2) y la 3 (7.499,8), pero también la más baja, la 6 (4.011,6). Sin las comunas 1 y 3, el promedio de la zona cae de 6.049 a 4.893, por debajo del promedio de CABA. En las comunas 1 y 3, con ingresos algo inferiores al promedio, el delito parece responder más a la oportunidad que a la vulnerabilidad: la concentración de comercios, oficinas, transporte y turismo atrae a muchas personas que no viven allí, lo que aumenta los posibles blancos de delito (hurtos y robos en la vía pública). Como esas personas no se cuentan en la población residente, la tasa de la Comuna 1 está sobreestimada.

**Zona Sur (Comunas 4, 8 y 9).** Concentra las peores condiciones de vida: en 2024, el ingreso per cápita familiar de la Comuna 8 era apenas el 51 % del promedio de la Ciudad, y el de las comunas 4 y 9, el 73 % y el 77 %. También tiene la mayor desocupación (10,0 %, 9,3 % y 6,9 %) y un hacinamiento alto en las comunas 8 (17,9 %) y 4 (15,7 %). Es la zona con la tasa promedio más alta (5.952), pero no es homogénea: la Comuna 4 (6.920,6) y la Comuna 9 (5.959,0) superan el promedio de la Ciudad, mientras que la Comuna 8, la más pobre, queda por debajo (4.976,5). Por eso, la vulnerabilidad no puede leerse como causa directa del nivel de delito. Entre las explicaciones posibles figuran una menor propensión a denunciar, menos blancos de oportunidad y diferencias en el tipo de delito que la tasa total no muestra.

**Zona Norte (Comunas 2, 11, 12, 13 y 14).** Tiene los ingresos más altos (en 2024, las comunas 2, 14 y 13 superaban en un 23 % a 33 % el promedio de la Ciudad), la desocupación más baja y un hacinamiento mínimo. Es la zona con la tasa promedio más baja (4.815), pero tampoco es homogénea: las comunas 11, 12 y 13 están entre las de menor delito, mientras que la Comuna 14 (6.154,8) supera el promedio de la Ciudad. Su posición se explica por Palermo, un polo gastronómico, comercial y turístico. Que un mayor ingreso no siempre implique menos delito confirma que los delitos contra la propiedad también responden a la oportunidad (riqueza expuesta y tránsito de personas), no solo a la privación.


## 6. Condiciones territoriales de las comunas
Esta sección complementa el análisis de las condiciones de vida de los hogares con indicadores del territorio: la presencia de barrios populares, las condiciones habitacionales y educativas relevadas por el Censo 2022 y la estructura comercial (usos del suelo). El objetivo es caracterizar cada comuna para luego relacionar estas condiciones con la frecuencia y el tipo de delitos registrados.


### 6.1 Datos utilizados

Se caracteriza a las 15 comunas de la Ciudad con dos fuentes:
- **Censo Nacional 2022 (INDEC):** cobertura de salud, nivel educativo,
  tipo de vivienda y régimen de tenencia.
- **Registro de barrios populares (Gobierno de la Ciudad, BA Data):** cantidad y superficie
  de villas, asentamientos y conjuntos de vivienda social.

Los indicadores se calcularon como porcentajes para que las comunas sean comparables,
y cada barrio popular se asignó a su comuna según su ubicación en el mapa oficial de comunas.

### 6.2 Fuentes de datos

**Censo Nacional de Población, Hogares y Viviendas 2022 (INDEC)**, cuadros por comuna de la Ciudad de Buenos Aires
([resultados definitivos](https://www.indec.gob.ar/indec/web/Nivel4-Tema-2-41-165)):
- Salud y previsión social: población por tipo de cobertura de salud.

- Viviendas: viviendas particulares por tipo de vivienda.
- Condiciones habitacionales de la población: población por régimen de tenencia de la vivienda.

**Sobre los años de los datos:** el Censo 2022 es la fuente más reciente con información por comuna sobre salud, educación, vivienda y tenencia, ya que la EAH no releva estos indicadores con ese nivel de detalle y el próximo censo se realizará recién hacia 2032. Si bien el resto del trabajo cubre el período 2017-2025, 2022 se ubica dentro de ese período y las condiciones que mide el censo son estructurales: cambian lentamente y mantienen el orden entre comunas de un año a otro. Por eso consideramos válido usarlas como una "foto" de las condiciones territoriales de las comunas para compararlas con el delito del período.

**Gobierno de la Ciudad de Buenos Aires (BA Data):**

- [Relevamiento de Usos del Suelo 2022-2024](https://data.buenosaires.gob.ar/dataset/relevamiento-usos-suelo): uso y estado de cada parcela.

**Construcción de la tabla:** la tabla `territorio_caba_2022.csv` se elaboró a partir de los cuadros del Censo

### 6.3 Estructura comercial de las comunas

Dentro de la investigación queremos relacionar el delito con la estructura comercial del territorio.
Para medirla se usa el **Relevamiento de Usos del Suelo 2022-2024** del Gobierno de la Ciudad (BA Data),
que registra el uso de cada parcela de la Ciudad: vivienda, comercio, oficinas, industria, etc.,
y si se encontraba activa o inactiva al momento del relevamiento.

A partir de este registro se calculan tres indicadores por comuna:
- **Locales comerciales activos:** cantidad de locales en funcionamiento (incluye galerías, shoppings y mercados).
- **% de locales vacíos:** proporción de locales comerciales inactivos. Es un indicador del dinamismo
  o del deterioro de la actividad económica de la zona.

### 6.6 Integración: condiciones de vida, territorio y tipo de delito

Las secciones 5 y 6 mostraron que el nivel de delito no se explica solo por el ingreso: la Comuna 8, la más vulnerable, tiene una tasa total por debajo del promedio. Para entender este resultado, separamos los delitos en dos grupos: **contra la propiedad** (robos, hurtos y sus tentativas) y **violentos contra las personas** (lesiones dolosas, amenazas y homicidios dolosos), y calculamos qué porcentaje representan los violentos en cada comuna.

```python
# 0. Colores por zona (se definen acá porque la variable "colores" se usó antes con otro contenido)
colores = {"Norte": "#EC6A9C", "Centro": "#24A57A", "Sur": "#3A8FE0"}

# 1. Leemos los delitos del SNIC y nos quedamos con las comunas de CABA
delitos = pd.read_csv(url_datos + "snic_amba_2017_2025.csv")
delitos = delitos[delitos["departamento_nombre"].str.startswith("Comuna", na=False)].copy()
delitos["comuna"] = delitos["departamento_nombre"].str.replace("Comuna ", "").astype(int)

# 2. Agrupamos los delitos en dos tipos
contra_propiedad = ["Robos (excluye los agravados por el resultado de lesiones y/o muertes)",
                    "Tentativas de robo (excluye las agravadas por el res. de lesiones y/o muerte)",
                    "Hurtos", "Tentativas de hurto"]
violentos = ["Lesiones dolosas", "Amenazas", "Homicidios dolosos"]

# 3. Tasa promedio anual (2017-2025) de cada tipo, por comuna
prop = delitos[delitos["codigo_delito_snic_nombre"].isin(contra_propiedad)]
prop = prop.groupby(["comuna", "anio"])["tasa_hechos"].sum().groupby("comuna").mean()
viol = delitos[delitos["codigo_delito_snic_nombre"].isin(violentos)]
viol = viol.groupby(["comuna", "anio"])["tasa_hechos"].sum().groupby("comuna").mean()

tipo_delito = pd.DataFrame({"tasa_propiedad": prop, "tasa_violentos": viol}).reset_index()
tipo_delito["% delitos violentos"] = (tipo_delito["tasa_violentos"] /
                                      (tipo_delito["tasa_propiedad"] + tipo_delito["tasa_violentos"]) * 100).round(1)

# 4. Lo sumamos a mi tabla territorial (si ya estaban, las reemplazamos)
df_territorio = df_territorio.drop(columns=["tasa_propiedad", "tasa_violentos", "% delitos violentos"], errors="ignore")
df_territorio = df_territorio.merge(tipo_delito, on="comuna")

# 5. Gráfico de barras: % de delitos violentos por comuna, coloreado por zona
ordenado = df_territorio.sort_values("% delitos violentos")
fig, ax = plt.subplots(figsize=(9, 6))
barras = ax.barh("Comuna " + ordenado["comuna"].astype(str), ordenado["% delitos violentos"],
                 color=ordenado["zona"].map(colores))
ax.bar_label(barras, fmt="%.1f%%", padding=3, fontsize=9)
ax.set_title("Peso de los delitos violentos sobre el total por comuna\n(CABA, promedio 2017-2025)",
             fontsize=12, fontweight="bold")
ax.set_xlabel("% de delitos violentos (lesiones, amenazas y homicidios)")
ax.set_ylabel("Comuna")
ax.legend(handles=[Patch(color=colores[z], label=z) for z in ["Norte", "Centro", "Sur"]],
          title="Zona", loc="lower right")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", visible=False)
ax.set_xlim(0, 30)
plt.tight_layout()
plt.savefig("grafico_delitos_violentos.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Interpretación:** el peso de los delitos violentos crece de norte a sur. Las tres comunas de la Zona Sur encabezan el ranking, con la Comuna 8 en primer lugar (25,9 %), mientras que las comunas del Norte, como la 14 (10,0 %), la 13 (10,9 %) y la 2 (11,0 %), registran los valores más bajos. Este patrón es consistente con la hipótesis de vulnerabilidad: la violencia tiene mayor presencia relativa en las comunas con peores condiciones sociales.

Los delitos violentos representan el 21,6 % del total en la Zona Sur, frente al 14,7 % en el Centro y el 12,3 % en el Norte. Las comunas con más personas sin obra social y más barrios populares son las que registran mayor peso de lesiones, amenazas y homicidios, con la Comuna 8 a la cabeza (25,9 %).

**Hipótesis de deterioro (parcialmente confirmada).** Las comunas con más locales vacíos tienden a tener mayor peso de delitos violentos: la Comuna 4 combina la mayor proporción de locales vacíos de la Ciudad (29,3 %) con un 21,9 % de delitos violentos, y la Zona Sur encabeza ambos indicadores. Sin embargo, los locales vacíos se concentran en las mismas comunas más vulnerables, por lo que no es posible distinguir si la violencia se asocia al deterioro comercial en sí o a las condiciones sociales que lo acompañan. Además, hay excepciones, como la Comuna 11, que tiene muchos locales vacíos (24,0 %) pero un peso de delitos violentos intermedio (15,3 %).

## 7. Conclusión general

**Comparación entre CABA y el Conurbano.** El Conurbano registra más hechos delictivos que la Ciudad en todo el período, por su mayor población. Pero al considerar la tasa cada 100.000 habitantes, la relación se invierte: las comunas porteñas promedian unos 5.620 hechos anuales, alrededor del doble que los partidos del Conurbano (unos 2.720), y aun la comuna con menor tasa supera al partido con la tasa más alta.

**Delito y condiciones socioeconómicas en las comunas porteñas.** Dentro de la Ciudad, las condiciones de vida y del territorio empeoran de norte a sur: el Sur concentra los menores ingresos, la mayor desocupación, la menor cobertura de salud, la mayoría de los barrios populares y la mayor proporción de locales vacíos. Sin embargo, estas condiciones no determinan **cuánto** delito se registra, sino **qué tipo** de delito predomina. Los delitos contra la propiedad se concentran donde hay más actividad comercial y circulación de personas, como las comunas 1 y 3, mientras que los delitos violentos tienen mayor peso en las comunas más vulnerables del Sur. La Comuna 8 resume este hallazgo: tiene una tasa total por debajo del promedio, pero la mayor proporción de delitos violentos de la Ciudad.

**Limitaciones.** Los resultados se refieren a comunas, no a personas: no permiten afirmar que los residentes de una comuna con peores condiciones delinquen más o son más victimizados. Tampoco demuestran causalidad, ya que hay factores que no se miden (población flotante, propensión a denunciar, patrullaje) y el análisis se apoya en solo 15 comunas. Se trata de delitos denunciados, que dependen de la decisión de denunciar y pueden subestimar los hechos en algunos territorios. Por último, el análisis combina fuentes de distintos años: las condiciones territoriales corresponden al Censo 2022, mientras que el delito abarca 2017-2025; aunque son condiciones estructurales que varían lentamente, no reflejan los cambios posteriores a 2022.
